Experiment 6: Statistical Distribution Analysis of Urban Climate Data

Aim: To analyze the distributions of temperature, humidity, and precipitation using statistics and graphs.

Dataset: `urban_climate.csv`  

## 1. Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("urban_climate.csv")
df.head()
```

## 2. Select suitable attributes
The selected attributes are temperature, humidity, and precipitation.

```python
attributes = ["temperature_celsius", "humidity_percent", "precipitation_mm"]
df[attributes].describe()
```

## 3. Calculate mean, median, and standard deviation

**Observation:** Mean and median describe the central value. Standard deviation describes how spread out the readings are. Record and compare the printed results.

## 4. Plot histograms

```python
for column in attributes:
    data = df[column].dropna()
    plt.hist(data, bins=20, edgecolor="black")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title("Histogram of " + column)
    plt.show()
```

**Observation:** Describe where values are concentrated and whether each histogram is roughly symmetric, skewed, or has multiple peaks.

## 5. Generate density plots

```python
for column in attributes:
    data = df[column].dropna()
    plt.figure()
    data.plot(kind="density")
    plt.xlabel(column)
    plt.title("Density plot of " + column)
    plt.show()
```

**Observation:** A density plot is a smoothed view of the distribution. Peaks indicate ranges where observations are more concentrated.

## 6. Calculate skewness and kurtosis

```python
for column in attributes:
    data = df[column].dropna()
    print("\\n", column)
    print("Skewness:", data.skew())
    print("Excess kurtosis:", data.kurt())
```

**Earlier calculated observations:**
- Temperature: skewness ≈ 0.085, excess kurtosis ≈ −0.488 — nearly symmetric, slight right skew, lighter tails.
- Humidity: skewness ≈ 0.074, excess kurtosis ≈ −0.398 — nearly symmetric, slight right skew, lighter tails.
- Precipitation: skewness ≈ 0.790, excess kurtosis ≈ 0.105 — right-skewed, with excess kurtosis close to zero.

These are the previously calculated moment-based values. Pandas may report slightly different values because it uses adjusted sample formulas.

## 7. Identify potential outliers using the IQR rule

```python
for column in attributes:
    data = df[column].dropna()
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = data[(data < lower) | (data > upper)]

    print("\\n", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Potential outlier count:", len(outliers))
```

**Observation:** Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are flagged as potential outliers. Your earlier temperature result was Q1 = 2.81, Q3 = 23.85, limits −28.75 to 55.41, and zero potential outliers. Note the humidity and precipitation counts printed by the code.

## 8. Compare observed data with a theoretical distribution
Overlay a normal distribution curve on each histogram. The curve uses the variable's mean and standard deviation.

```python
for column in attributes:
    data = df[column].dropna()
    mean = data.mean()
    sd = data.std()

    plt.hist(data, bins=20, density=True, alpha=0.6, edgecolor="black")
    x = np.linspace(data.min(), data.max(), 200)
    curve = (1 / (sd * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / sd) ** 2)
    plt.plot(x, curve)
    plt.xlabel(column)
    plt.ylabel("Density")
    plt.title("Observed data and normal curve: " + column)
    plt.show()
```

**Observation:** If the histogram bars roughly follow the bell-shaped curve, a normal distribution may be a reasonable approximation. A mismatch suggests that the normal model may not describe the data well. Inspect each graph, especially precipitation, for visible skew.
